## Bài 5

```python
def tim_vi_tri_k(lst, k):
    for i in range(len(lst)):
        if lst[i] == k:
            return i 
    return -1
s=input()
mang_so=[int(x) for x in s.split()]
so_can_tim=int(input())
ket_qua=tim_vi_tri_k(mang_so,so_can_tim)
print(f"vị trí đầu tiên của {so_can_tim} là {ket_qua}")    
```

## Bài 9

```python
def push_zeros_to_front(arr):
    zeros = [x for x in arr if x == 0]
    others = [x for x in arr if x != 0]
    return zeros + others
arr =list(map(int,input().split()))
result = push_zeros_to_front(arr)
print("Đẩy 0 về đầu:", result)
```

## Bài 13

```python
def check_sequence(arr):
    is_increasing = True
    is_decreasing = True

    for i in range(len(arr) - 1):
        if arr[i] >= arr[i+1]: 
            is_increasing = False
        if arr[i] <= arr[i+1]: 
            is_decreasing = False

    if is_increasing:
        return "Tang thuc su"
    elif is_decreasing:
        return "Giam thuc su"
    else:
        return "Vo danh"
arr = list(map(int, input("Nhap day so (cach nhau boi khoang trang): ").split()))
print("Ket qua:", check_sequence(arr))

    
```
