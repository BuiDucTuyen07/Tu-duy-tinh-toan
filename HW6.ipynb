```python
# Nhập số hàng n và số cột m
n = int(input("Nhap so hang n: "))
m = int(input("Nhap so cot m: "))

# Khởi tạo mảng rỗng
matrix = []

# Nhập các phần tử của mảng
print(f"Nhap cac phan tu cho ma tran {n}x{m}:")
for i in range(n):
    # Nhập cả một hàng, cách nhau bởi dấu cách và chuyển thành list số nguyên
    row = list(map(int, input(f"Nhap hang {i+1}: ").split()))
    matrix.append(row)

# In lại mảng ra màn hình dưới dạng ma trận
print("\nMa tran ket qua:")
for row in matrix:
    for element in row:
        # Sử dụng f-string với định dạng :4d để căn phải 4 vị trí
        print(f"{element:4d}", end="")
    print() # Xuống dòng sau khi in xong một hàng
```
