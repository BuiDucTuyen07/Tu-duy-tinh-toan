## Bài 1

```python
def is_amstrong(x:int)->bool:
    if x<0:
        return False
    str_x=str(x)
    n=len(str_x)
    total_sum=sum(int(digit)**n for digit in str_x)
    return total_sum==x
number=int(input())
if is_amstrong(number):
    print(f"{number} là số amstrong")
else:
    print(f"{number} ko phải số amstrong")
```

## Bài 2

```python
def la_so_chinh_phuong(x:int)->bool:
    if x<0:
        return False
    root=int(x**(1/2))
    return root*root==x
x=int(input())
if la_so_chinh_phuong(x):
    print(f"{x} là số chính phương")
else:
    print(f"{x} ko phải số chính phương")
```

## Bài 3

```python
def la_so_doi_xung(x:int)->bool:
    if x<0:
        return False
    so_ban_dau=x
    so_dao_nguoc=0
    while x>0:
        chu_so_cuoi=x%10
        so_dao_nguoc=so_dao_nguoc*10+chu_so_cuoi
        x=x//10
    return so_ban_dau==so_dao_nguoc
n=int(input())
if la_so_doi_xung(x):
        print(f"{n} là số đối xứng")
else:
        print(f"{n} ko phải số đối xứng")
    
```

## Bài 4

```python
def la_so_ky_dieu(x:int)->bool:
    if x<=0:
        return False
    while x>=10:
        tong_chu_so=0
        while x>0:
            tong_chu_so += x%10
            x //= 10
        x=tong_chu_so
    return x==1
x=int(input())
if la_so_ky_dieu(x):
    print(f"{x} là số kỳ diệu")
else:
    print(f"{x} ko là số kỳ diệu")
```

## Bài 5

```python
def detect_number_type(s: str) -> str:
    try:
        int(s)
        return 'int'
    except ValueError:
        pass
    try:
        float(s)
        return 'float'
    except ValueError:
        pass
    return 'neither'
s=input()
print(f"{s}-> {detect_number_type(s)}")
```

```python
def sum_halves(s:str)->str:
    first_half=s[0:3]
    second_half=s[3:6]
    return int(first_half)+int(second_half)
chuoi_nhap=input()
if len(chuoi_nhap) == 6 and chuoi_nhap.isdigit():
    ket_qua = sum_halves(chuoi_nhap)
    print(f"Tổng hai nửa của {chuoi_nhap} là: {ket_qua}")
else:
    print("Vui lòng nhập đúng 6 chữ số!")
```

```python
def quadratic_roots_count(a,b,c):
    if a==0:
        if b==0:
            return -1 if c==0 else 0
        return 1
    delta=b**2-4*a*c
    if delta>0:
        return 2
    elif delta==0:
        return 1
    else:
        return 0
a,b,c=map(int,input().split())
print(f"{a}x^2+{b}x+{c} có {quadratic_roots_count(a,b,c)} nghiệm")    
```

## Bài 9

```python
def are_collinear(x1, y1, x2, y2, x3, y3):
    return (x2 - x1) * (y3 - y1) == (x3 - x1) * (y2 - y1)
x1, y1, x2, y2, x3, y3=map(int,input().split())
if are_collinear(x1, y1, x2, y2, x3, y3):
    print(f"3 điểm {x1,y1} và {x2,y2} và {x3,y3} là 3 điểm thẳng hàng")
else:
    print(f"3 điểm {x1,y1} và {x2,y2} và {x3,y3} là 3 điểm ko thẳng hàng")
```

## Bài 12

```python
def count_non_zero_digits(numbers_list):
    count = 0
    for num in numbers_list:
        s = str(abs(num))
        for char in s:
            if char != '0':
                count += 1
    return count
try:
    input_data = input().split()
    my_list = [int(x) for x in input_data]
    result = count_non_zero_digits(my_list)
    print(f"Số lượng chữ số khác 0 là: {result}")

except ValueError:
    print("Vui lòng chỉ nhập các số nguyên!")
```

```python
def get_max_min_manual(a, b, c):
    v_max = a
    v_min = a
    if b > v_max: v_max = b
    if c > v_max: v_max = c
    if b < v_min: v_min = b
    if c < v_min: v_min = c
    
    return [v_max, v_min]
a,b,c=map(float,input().split())
ket_qua=get_max_min_manual(a, b, c)
print(f"max là {ket_qua[0]},min là {ket_qua[1]}")
```

## Bài 15

```python
import math
def binomial_coefficient(n, k):
    if k < 0 or k > n:
        return 0
    return math.comb(n, k)
n = int(input())
k = int(input())
print(f"C({n}, {k}) = {binomial_coefficient(n, k)}")
```

## Bài 16

```python
def sum_of_squares(n):
    tong = 0
    for i in range(1, n + 1):
        tong += i**2
    return tong
n = int(input())
print(f"Tổng bình phương từ 1 đến {n} là: {sum_of_squares(n)}")
```

## Bài 18

```python
def sum_1_to_n(n):
    if n<=0:
        return 0
    tong=0
    for x in range(1,n+1):
        s_x=str(x)
        if len(set(s_x))==len(s_x):
            tong+=x
    return tong
n=int(input())
print(f"tổng các số thỏa mãn đến {n} là { sum_1_to_n(n)}")
```
